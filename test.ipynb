```python
import requests
import pandas as pd

# Ваш API-ключ USDA
API_KEY = "pyUyaJ5S7B3aTS36Q022CeuLxBuvO4S2eVuGaDFA"
BASE_URL = "https://api.nal.usda.gov/fdc/v1"

# Функция для поиска продуктов
def search_food(query):
    url = f"{BASE_URL}/foods/search"
    params = {
        "api_key": API_KEY,
        "query": query
    }
    response = requests.get(url, params=params)
    return response.json()

# Функция для получения деталей продукта
def get_food_details(fdc_id):
    url = f"{BASE_URL}/food/{fdc_id}"
    params = {
        "api_key": API_KEY
    }
    response = requests.get(url, params=params)
    return response.json()

# Функция для преобразования данных в датафрейм
def parse_nutrients_to_df(food_details):
    nutrients = food_details.get("foodNutrients", [])
    data = []
    for nutrient in nutrients:
        row = {
            "Nutrient ID": nutrient.get("nutrientId"),
            "Nutrient Name": nutrient.get("nutrientName"),
            "Unit": nutrient.get("unitName"),
            "Value": nutrient.get("value")
        }
        data.append(row)
    return pd.DataFrame(data)

# Пример использования
if __name__ == "__main__":
    # Поиск продукта
    query = "herring"
    search_results = search_food(query)
    
    if search_results.get("foods"):
        # Получение ID первого найденного продукта
        fdc_id = search_results["foods"][0]["fdcId"]
        
        # Получение деталей продукта
        food_details = get_food_details(fdc_id)
        
        # Преобразование данных в датафрейм
        df = parse_nutrients_to_df(food_details)
        
        # Вывод датафрейма
        print(f"Данные для продукта: {food_details.get('description')}")
        display(df)
```

```python
def max_sq(min_size: int, max_size: int, steps=1):
    """
    Функция находит максимальный размер квадрата, на которые делится без 
    остатка прямоугольный участок с переданными размерами

    Args:
        min_size (int): меньший размер участка
        max_size (int): больший размер участка
        steps (int, optional): для подсчета шагов.

    Returns:
        int: Сторона наибольшего квадрата на которые делится участок без остатка
        int: Количество шагов в решении
    """

    if max_size % min_size == 0:
        return min_size, steps
    else:
        new_max_size = min_size
        new_min_size = max_size % min_size
        return max_sq(new_min_size, new_max_size, steps=steps + 1)


min_side_size = int(input('Введите меньшую сторону участка:'))
max_side_size = int(input('Введите большую сторону участка:'))

side_size, steps = max_sq(min_side_size, max_side_size)

print(f'Наибольшая сторона квадрата - {side_size}, шагов понадобилось - {
    steps}')
```

```python
import re

def acronym(some_words):
    result = ''
    str_ = some_words.upper()
    lst_ = str_.split(' ')
    # result = [i[0] for i in lst_]
    # for i in lst_:
    #     result += i[0]
    result = ''.join([i[0] for i in lst_])
    return result

print(acronym('Информационные технологии'))
print(acronym('Near Field Communication'))
```

```python
import re
 
emails = ['test@gmail.com', 'xyz@test.in', 'test@ya.ru', 'xyz@mail.ru', 'xyz@ya.ru', 'xyz@gmail.com']
 
def counting(emails: list) -> dict:
    dic = {}
    for item in emails:
        match = re.search(r"@\w+\.\w+", item)
        if match:
            domain = match.group()[1:]
            dic.setdefault(domain, 0)
            dic[domain] += 1
    return dic
 
for k, v in counting(emails).items():
    print(k, v)
    
str(emails)
```
